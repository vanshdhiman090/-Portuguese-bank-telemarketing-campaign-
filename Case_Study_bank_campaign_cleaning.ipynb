# Ask Phase

**Business Problem:**

*A Portuguese bank conducted a telemarketing campaign to sell term deposit subscriptions to existing customers. The campaign generated 45,211 customer interactions. Leadership needs to understand which customer profiles and campaign strategies drove conversions — and which wasted resources.*

**Goal:**

*Identify the key drivers of term deposit subscription using customer demographic and campaign interaction data. Deliver actionable recommendations to improve conversion rate and reduce cost-per-acquisition in future campaigns.*

**Stakeholder:**

*Marketing Director and Campaign Strategy Team.*

**Success Metric:**

*A dashboard that clearly identifies high-converting customer segments and campaign behaviours — with at least 3 actionable recommendations.*

**Business Questions:**


*   What was the overall conversion rate of the campaign — and is it acceptable?

*   Which customer segments (age, job, education) converted at the highest rate?


*   Does call duration and number of contacts impact conversion?

*   Did customers with a successful previous campaign convert again?


*   Which months and contact methods drove the best results?

# Analysis Phase

```python
# @title
# Overall conversion rate
total = len(df)
converted = df['subscribed_num'].sum()
not_converted = total - converted
rate = round(df['subscribed_num'].mean() * 100, 2)

print(f"Total customers contacted: {total}")
print(f"Converted: {converted}")
print(f"Not converted: {not_converted}")
print(f"Conversion rate: {rate}%")
```

```python
# @title
# Conversion rate by job
job_conv = df.groupby('job')['subscribed_num'].agg(['mean', 'count'])
job_conv['mean'] = round(job_conv['mean'] * 100, 2)
job_conv.columns = ['conversion_rate_%', 'total_contacts']
job_conv = job_conv.sort_values('conversion_rate_%', ascending=False)
print("=== CONVERSION BY JOB ===")
print(job_conv)

# Conversion rate by education
edu_conv = df.groupby('education')['subscribed_num'].agg(['mean', 'count'])
edu_conv['mean'] = round(edu_conv['mean'] * 100, 2)
edu_conv.columns = ['conversion_rate_%', 'total_contacts']
edu_conv = edu_conv.sort_values('conversion_rate_%', ascending=False)
print("\n=== CONVERSION BY EDUCATION ===")
print(edu_conv)

# Conversion rate by age group
age_conv = df.groupby('age_group')['subscribed_num'].agg(['mean', 'count'])
age_conv['mean'] = round(age_conv['mean'] * 100, 2)
age_conv.columns = ['conversion_rate_%', 'total_contacts']
print("\n=== CONVERSION BY AGE GROUP ===")
print(age_conv)

# Conversion rate by marital status
mar_conv = df.groupby('marital')['subscribed_num'].agg(['mean', 'count'])
mar_conv['mean'] = round(mar_conv['mean'] * 100, 2)
mar_conv.columns = ['conversion_rate_%', 'total_contacts']
mar_conv = mar_conv.sort_values('conversion_rate_%', ascending=False)
print("\n=== CONVERSION BY MARITAL STATUS ===")
print(mar_conv)
```

```python
# @title
# Conversion rate by number of contacts
contact_conv = df.groupby('contacts_this_campaign')['subscribed_num'].agg(['mean', 'count'])
contact_conv['mean'] = round(contact_conv['mean'] * 100, 2)
contact_conv.columns = ['conversion_rate_%', 'total_contacts']
print("=== CONVERSION BY NUMBER OF CONTACTS ===")
print(contact_conv.head(10))

# Average call duration — converted vs not
duration_analysis = df.groupby('subscribed')['call_duration_min'].agg(['mean', 'median'])
duration_analysis.columns = ['avg_duration_min', 'median_duration_min']
print("\n=== CALL DURATION VS CONVERSION ===")
print(round(duration_analysis, 2))

# Conversion by contact method
contact_method = df.groupby('contact')['subscribed_num'].agg(['mean', 'count'])
contact_method['mean'] = round(contact_method['mean'] * 100, 2)
contact_method.columns = ['conversion_rate_%', 'total_contacts']
print("\n=== CONVERSION BY CONTACT METHOD ===")
print(contact_method)
```

```python
# @title
# Conversion by month
month_conv = df.groupby(['month_num', 'month'])['subscribed_num'].agg(['mean', 'count'])
month_conv['mean'] = round(month_conv['mean'] * 100, 2)
month_conv.columns = ['conversion_rate_%', 'total_contacts']
month_conv = month_conv.sort_values('month_num')
print("=== CONVERSION BY MONTH ===")
print(month_conv)

# Conversion by day of week
day_conv = df.groupby('contact_day')['subscribed_num'].agg(['mean', 'count'])
day_conv['mean'] = round(day_conv['mean'] * 100, 2)
day_conv.columns = ['conversion_rate_%', 'total_contacts']
print("\n=== CONVERSION BY DAY OF WEEK ===")
print(day_conv)
```

```python
# Housing loan vs conversion
housing_conv = df.groupby('housing')['subscribed_num'].agg(['mean', 'count'])
housing_conv['mean'] = round(housing_conv['mean'] * 100, 2)
housing_conv.columns = ['conversion_rate_%', 'total_contacts']
print("\n=== HOUSING LOAN VS CONVERSION ===")
print(housing_conv)

# Personal loan vs conversion
loan_conv = df.groupby('loan')['subscribed_num'].agg(['mean', 'count'])
loan_conv['mean'] = round(loan_conv['mean'] * 100, 2)
loan_conv.columns = ['conversion_rate_%', 'total_contacts']
print("\n=== PERSONAL LOAN VS CONVERSION ===")
print(loan_conv)
```

2.  Retired and student segments are highest performing.

3. Conversion drops sharply after 5+ contact attempts.
   Diminishing returns are significant — contacts beyond 5 waste resources.

4. Cellular contact outperforms telephone by around 10 percentage.

5. March , December and September show highest conversion rates.
   May has highest volume but lowest conversion — misaligned effort.

### Executive Summary

SITUATION
41,188 customers contacted. Overall conversion rate: 11.27%

RECOMMENDATION 1 — PRIORITIZE BOOKEND DEMOGRAPHICS
Target 60+ and student segments. Suppress outreach to
blue-collar and services unless prior engagement exists.

RECOMMENDATION 2 — IMPLEMENT 5-CALL HARD CAP
Cap outbound calls at 5 attempts per customer. Every call
beyond that inflates CPA with near-zero conversion return.

RECOMMENDATION 3 — MOBILE-FIRST STRATEGY
Prioritize cellular numbers in auto-dialer before landlines.
This alone could lift conversion without adding new contacts.

RECOMMENDATION 4 — WARM LEAD ROUTING
Route previously engaged customers to senior agents first.
These leads convert at up to 60% — highest value segment.

PROJECTED IMPACT
Implementing all four recommendations on a future 10,000
customer campaign could lift conversion from 11.27% to
18–22% while reducing cost-per-acquisition by 30–40%.
